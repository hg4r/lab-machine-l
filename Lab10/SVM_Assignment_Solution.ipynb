# Support Vector Machines – Iris Dataset

Welcome to the SVM Assignment! We will be analyzing the famous **Iris flower dataset** using Support Vector Machines.

## The Data

The [Iris flower data set](http://en.wikipedia.org/wiki/Iris_flower_data_set) was introduced by Sir Ronald Fisher in 1936. It consists of 150 samples from three species of Iris flowers:

- **Iris setosa**
- **Iris versicolor**  
- **Iris virginica**

Four features were measured from each sample:

| Feature | Description |
|---|---|
| sepal_length | Sepal length in cm |
| sepal_width | Sepal width in cm |
| petal_length | Petal length in cm |
| petal_width | Petal width in cm |
| species | Flower species (target) |

## Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

## Get the Data

**Load the iris dataset using sklearn (since seaborn requires internet). We'll structure it exactly like sns.load_dataset('iris').**

```python
from sklearn.datasets import load_iris

iris_sk = load_iris()
iris = pd.DataFrame(iris_sk.data, columns=['sepal_length', 'sepal_width', 'petal_length', 'petal_width'])
iris['species'] = pd.Categorical.from_codes(iris_sk.target, iris_sk.target_names)

print(iris.shape)
iris.head()
```

```python
iris.info()
```

```python
iris.describe()
```

## Exploratory Data Analysis

Time to put your data visualization skills to the test!

**Create a pairplot of the dataset. Which flower species seems to be the most separable?**

```python
sns.pairplot(iris, hue='species', palette='Set1')
plt.suptitle('Iris Dataset - Pairplot by Species', y=1.02)
plt.show()

# Answer: Iris setosa is the most separable — it forms a clearly distinct cluster
# from versicolor and virginica in nearly every feature combination.
```

**Create a KDE plot of sepal_length versus sepal_width for the setosa species.**

```python
setosa = iris[iris['species'] == 'setosa']

sns.kdeplot(data=setosa, x='sepal_length', y='sepal_width',
            fill=True, cmap='Blues', thresh=0.05)
plt.title('KDE Plot: Sepal Length vs Sepal Width (Setosa)')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.show()
```

## Train Test Split

**Split the data into a training set and a testing set.**

```python
from sklearn.model_selection import train_test_split

X = iris.drop('species', axis=1)
y = iris['species']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=101)

print('Training set size:', X_train.shape)
print('Testing set size: ', X_test.shape)
```

## Train a Model

**Call the SVC() model from sklearn and fit the model to the training data.**

```python
from sklearn.svm import SVC

model = SVC()
model.fit(X_train, y_train)
```

## Model Evaluation

**Get predictions from the model and create a confusion matrix and a classification report.**

```python
from sklearn.metrics import classification_report, confusion_matrix

predictions = model.predict(X_test)
```

```python
print('Confusion Matrix:')
print(confusion_matrix(y_test, predictions))
```

```python
print('Classification Report:')
print(classification_report(y_test, predictions))
```

> **Note:** The model performs very well (98% accuracy) even without any parameter tuning! The Iris dataset is relatively simple and well-structured, which makes SVM a great choice here.

## Gridsearch Practice

Let's see if we can tune the parameters to get even better results using **GridSearchCV**.

**Import GridSearchCV from Scikit-Learn.**

```python
from sklearn.model_selection import GridSearchCV
```

**Create a dictionary called `param_grid` with values for C and gamma to search over.**

```python
param_grid = {'C': [0.1, 1, 10, 100],
              'gamma': [1, 0.1, 0.01, 0.001],
              'kernel': ['rbf']}
```

**Create a GridSearchCV object and fit it to the training data.**

```python
grid = GridSearchCV(SVC(), param_grid, verbose=3, cv=5)
grid.fit(X_train, y_train)
```

```python
# Best parameters found by GridSearch
print('Best Parameters:', grid.best_params_)
print('Best Estimator:', grid.best_estimator_)
```

**Now create predictions using the tuned grid model and show the classification report and confusion matrix.**

```python
grid_predictions = grid.predict(X_test)
```

```python
print('Confusion Matrix (GridSearch Tuned Model):')
print(confusion_matrix(y_test, grid_predictions))
```

```python
print('Classification Report (GridSearch Tuned Model):')
print(classification_report(y_test, grid_predictions))
```

## Were you able to improve?

**Answer:** In this case, GridSearch may not improve significantly over the default SVC because the Iris dataset is already well-suited for SVM and the default parameters happen to work well.

However, for more complex and noisy datasets (like the breast cancer dataset seen in the lecture), GridSearch can make a **significant** improvement — especially when the default SVC classifies everything into one class due to poor hyperparameters.

**Key Takeaways:**
- SVM works great for small, clean datasets like Iris
- GridSearchCV helps find the optimal `C` (controls margin hardness) and `gamma` (controls influence of individual training points)
- The `rbf` kernel is a powerful default choice for non-linear data
